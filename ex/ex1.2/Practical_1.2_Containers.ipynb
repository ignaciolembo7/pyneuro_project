## Exercise 1 — Excluding bad EEG trials

You performed EEG recordings from a subject in 10 trials total. Some trials have been marked as `"BAD"` due to eye blinks and poor electrode contact. Create a new list containing only the good trials.

Questions:
- What container type should `good_trials` be?
- Why is a list appropriate here?

```python
trial_ids = ["001", "002", "003", "004", "005",
             "006", "007", "008", "009", "010"]
bad_trials = ["004", "006", "007"]

good_trials = [trial for trial in trial_ids if trial not in bad_trials]
print(good_trials)
```

Aswers: 
- `good_trials` should be a list.

- A list is appropriate because preserves the order, can be indexed, and can be easily iterated over in later analysis steps.

## Exercise 2 — Trial filtering with two conditions

Starting from the same trial list, create a new list `usable_trials` that:
1. excludes all bad trials
2. excludes all trials with an odd trial number

The final result should only contain good **even-numbered** trials.

## Exercise 5 — append() vs extend() vs +

Create three separate lists starting from the same `base` list.

Create:
- `with_append` by adding `["temporal", "occipital"]` using `append()`
- `with_extend` by adding the same values using `extend()`
- `with_plus` by adding the same values using `+`

Questions:
- How does `with_append` differ from the other two?
- Which one contains a nested list?

```python
base = ["frontal", "parietal"]

with_append = base.copy()
with_append.append(["temporal", "occipital"])

with_extend = base.copy()
with_extend.extend(["temporal", "occipital"])

with_plus = base + ["temporal", "occipital"]

print(with_append)
print(with_extend)
print(with_plus)
```

Answers: 

- `with_append` contains a list within another list: `['frontal', 'parietal', ['temporal', 'occipital']]`. In contrast, `with_extend` and `with_plus` add the elements individually `['frontal', 'parietal', 'temporal', 'occipital']`

- The list that contains a nested list is `with_append`.

## Exercise 6 — Immutable vs mutable containers

Create:
- a list called `brain_regions_list`
- a tuple called `brain_regions_tuple`

Both should contain the same four strings:
`"frontal"`, `"parietal"`, `"temporal"`, `"occipital"`

Then:
1. change the second element of the list to `"cingulate"`
2. try to change the second element of the tuple in the same way

Answer:
- What happens in each case?
- Why?

```python
brain_regions_list = ["frontal", "parietal", "temporal", "occipital"]
brain_regions_tuple = ("frontal", "parietal", "temporal", "occipital")

brain_regions_list[1] = "cingulate"
print(brain_regions_list)

# Uncomment to test what happens with tuples:
# brain_regions_tuple[1] = "cingulate"
```

Answer: 
The list can be modified because lists are mutable objects. This means that we can change one of their elements after the list has been created. In this case, the second element of the list is replaced by "cingulate".

The tuple cannot be modified because tuples are immutable objects. Once a tuple is created, its elements cannot be reassigned. Trying to run brain_regions_tuple[1] = "cingulate" raises a TypeError, because tuples do not support item assignment.

Therefore, lists are useful when the collection may need to change during the analysis, while tuples are useful when the collection should remain fixed.

## Exercise 7 — Sets for unique labels

You have the following labels extracted from annotations.

Tasks:
1. convert `labels` into a set called `unique_labels`
2. print the set
3. print the number of unique labels
4. check whether `"blink"` is in the set
5. check whether `"line_noise"` is in the set

Then explain:
- Why is a set useful here?
- What information is lost when converting from list to set?

```python
labels = ["blink", "muscle", "blink", "good", "muscle", "good", "drift"]

unique_labels = set(labels)
print(unique_labels)
print(len(unique_labels))
print("blink" in unique_labels)
print("line_noise" in unique_labels)
```

Answer:

- Converting a list to a set loses the original order and also the number of times each label appeared.

## Exercise 12 — Copying and mutable objects

Run the first cell and inspect the output.

Then answer:
- Why did the dictionary change even though you only modified `a_list`?

After that, solve the problem by using `.copy()` so that changing the original list does **not** affect the copied object.

Finally, verify by printing both objects before and after modification.

```python
a_list = [0, 1]
a_dictionary = {"a": a_list}

print("Before change:", a_dictionary)

a_list[0] = 2

print("After change:", a_dictionary)

print("\nUsing copy():")

a_list = [0, 1]
another_list = a_list.copy()

print("Before:", a_list, another_list)
a_list[0] = 2
print("After:", a_list, another_list)
```

Answer: 

The dictionary changes because it doesn't store a separate copy of the list. It stores a reference to the same object, `a_list`. So, when you modify `a_list`, what you see in `a_dictionary["a"]` also changes.

With `.copy()`, `another_list` points to a new, separate list. That's why modifying `a_list` doesn't change `another_list`.

## Optional challenge 2 — Mixed-container reasoning

Predict the output before running.

Then explain:
- Why are `x` and `y` the same?
- Why is `z` different?

```python
x = [1, 2]
y = x
z = x.copy()

x.append(3)

print("x:", x)
print("y:", y)
print("z:", z)
```

Answer: 

x and y are equal because `y = x` doesn't create a copy, it makes `y` point to the same object as `x`. In contrast, `z = x.copy()` creates a new, independent list. That's why when we do `x.append(3)`, `y` also changes, but `z` doesn't.
