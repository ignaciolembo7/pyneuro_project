## Exercise 6 — Choose the better error-handling strategy

You want to convert user input to an integer.

**Option A**
```python
number = int(user_input)
print(number)
```

**Option B**
```python
try:
    number = int(user_input)
    print(number)
except ValueError:
    print("That is not a valid integer.")
```

Tasks:
1. Which option is safer when `user_input` may contain invalid text?
2. Give one example input for which Option A fails.
3. Explain why Option B is preferable in that case.

Answers:

1. Option B is better.

2. Option A fails when ``user_input`` cannot be converted to an integer. For example, if ``user_input = "hola" ``, ``int(user_input)`` raises a ``ValueError`` and the program stops.

3. Option B is preferable because, instead of crashing, the program catches the ``ValueError``, handles the invalid input and prints a message.

## Exercise 8 — Handle two different exception types

Test it with:
- `a = 10, b = 0`

## Exercise 10 — Debugging workflow

The following code contains **errors**.  
Run it, read the traceback, fix the error, and run again until the full code works.

````python
from math import squareroot

values = [4, 9, 16]
results = []

for i in range(4):
    results.append(squareroot(values[i]))

print("Square roots: " + results)
````

Questions:
1. Which error do you get first? 
2. After fixing it, which error appears next?
3. What is the final working version of the code?

Answers:

1. The first error is ``ImportError``, because squareroot is not a function in the math module.

2. After fixing the import, the next error is ``NameError`` if the code still calls squareroot instead of sqrt. After fixing that, an ``IndexError`` appears because range(4) tries to access values[3], but the list has only indices 0, 1, and 2. Finally, a ``TypeError`` appears if we try to concatenate a string and a list using +.

3. The final working version is:

````python
from math import sqrt

values = [4, 9, 16]
results = []

for i in range(3):
    results.append(sqrt(values[i]))

print("Square roots: ", results)
````

## Final reflection

Answer briefly:

1. What is the difference between a `SyntaxError` and an exception?______________________________________
2. Why is reading a traceback from bottom to top useful?_____________________________________________
3. Why is catching a specific exception usually better than writing a generic `except`?____________________________

Answers: 

1. A ``SyntaxError`` happens before the program runs, when Python cannot parse the code because its syntax is invalid. An ``exception`` happens during execution, when the code is syntactically valid but something goes wrong at runtime.

2. Reading a traceback from bottom to top is useful because the last line usually shows the exception type and the error message. The lines above show where the error occurred and how the program reached that point.

3. Catching a specific exception is usually better than using a generic except because it avoids hiding unexpected bugs. Specific exceptions make the code clearer and help ensure that only the expected error is handled.
